```python
#create datasets
import pandas as pd

df_old = pd.DataFrame({
    "customer_id": [101, 102, 103],
    "name": ["Alice", "Bob", "Charlie"],
    "amount": [100, 200, 300]
})

df_new = pd.DataFrame({
    "customer_id": [101, 102, 103],
    "name": ["Alice", "Bob", "Charlie"],
    "amount": [100, 250, 300]
})
```

```python
#first comparison
import datacompy

compare = datacompy.PandasCompare(
    df_old,
    df_new,
    join_columns=["customer_id"]
)
```

```python
print(compare.report())
```

```python
#using composite key 
orders_old = pd.DataFrame({
    "customer_id": [101, 101, 102],
    "order_id": [5001, 5002, 5003],
    "product": ["Laptop", "Mouse", "Keyboard"],
    "amount": [1000, 50, 80]
})

orders_new = pd.DataFrame({
    "customer_id": [101, 101, 102],
    "order_id": [5001, 5002, 5003],
    "product": ["Laptop", "Mouse", "Keyboard"],
    "amount": [1050, 50, 80]
})
```

```python
composite_key_results1 = datacompy.PandasCompare(
    orders_old,
    orders_new,
    join_columns=["customer_id"]
)
```

```python
print(composite_key_results1.report())
```

```python
composite_key_results2 = datacompy.PandasCompare(
    orders_old,
    orders_new,
    join_columns=["customer_id","order_id"]
)
print(composite_key_results2.report())
```

```python
#setting up tolerances 
sales_old = pd.DataFrame({
    "product_id": [1, 2, 3],
    "price": [100.00, 200.00, 300.00]
})

sales_new = pd.DataFrame({
    "product_id": [1, 2, 3],
    "price": [100.01, 200.03, 300.10]
})
```

```python
#this will show unequal values due to tolerance not set

compare = datacompy.PandasCompare(
    sales_old,
    sales_new,
    join_columns=["product_id"]
)

print(compare.report())
```

```python
compare = datacompy.PandasCompare(
    sales_old,
    sales_new,
    join_columns=["product_id"],
    abs_tol=0.05, #actual difference
    rel_tol=0.001 # for the number how much is the differences(relative tolernace)
)

print(compare.report())
```

Column Differences

```python
df1 = pd.DataFrame({
    "customer_id": [101, 102, 103],
    "name": ["Alice", "Bob", "Charlie"],
    "amount": [100, 200, 300]
})

df2 = pd.DataFrame({
    "customer_id": [101, 102, 103],
    "name": ["Alice", "Bob", "Charlie"],
    "amount": [100, 200, 400],
    "email": [
        "alice@example.com",
        "bob@example.com",
        "charlie@example.com"
    ]
})
```

```python
compare = datacompy.PandasCompare(
    df1,
    df2,
    join_columns=["customer_id"],
    
)

print(compare.report())
```

```python
print(f"matches(ignore_extra_columns=True): {compare.matches(ignore_extra_columns=True)}")
print(f"all_columns_match(): {compare.all_columns_match()}")
print(f"all_rows_overlap(): {compare.all_rows_overlap()}")
print(f"intersect_rows_match(): {compare.intersect_rows_match()}")
```

```python
#Find columns unique to each DataFrame
print(compare.df1_unq_columns())
print(compare.df2_unq_columns())
```

```python
print(compare.all_mismatch())
```
