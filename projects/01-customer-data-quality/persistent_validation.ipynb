```python
suite = context.suites.add(
    gx.ExpectationSuite(
        name="customer_quality_suite_v1"
    )
)
```

```python
#expectations

#customers should not be null

expectation = gx.expectations.ExpectColumnValuesToNotBeNull(
    column="customer_id"
)

suite.add_expectation(expectation)
```

```python
print(suite)
```

```python
email_regex = gx.expectations.ExpectColumnValuesToMatchRegex(
    column = 'email',
    regex = r"^[^@\s]+@[^@\s]+\.[^@\s]+$"

)

suite.add_expectation(email_regex)

email_not_null = gx.expectations.ExpectColumnValuesToNotBeNull(
    column='email'

)

suite.add_expectation(email_not_null)

age_between = gx.expectations.ExpectColumnValuesToBeBetween(
    column='age'
    ,min_value = 18
    ,max_value = 100

)

suite.add_expectation(age_between)
```

```python
expectation = gx.expectations.ExpectColumnValuesToBeUnique(
    column="customer_id"
)

suite.add_expectation(expectation)
```

```python
import pandas as pd
df = pd.read_csv(r'data\customers.csv')

#How many rows and columns does the dataset have?
print(df.shape) # 10 rows and 6 columns 

#Should customer_id be nullable? 
# no 
print(df.dtypes)

#Should customer_id be unique?
#yes 
print(df.isnull().sum())

#What values should status be allowed to contain?
#active and inactive 
print(df["customer_id"].nunique())

#What would be a reasonable validation rule for age?
#age should be between 10 and 100 

print(df["country"].value_counts())
print(df["status"].value_counts())
```

```python
#add data source
data_source = context.data_sources.add_pandas(name='customer_data_source')
print(data_source)
```

```python
#data asset 

data_asset = data_source.add_dataframe_asset(name='customer')
```

```python
#batch definition
batch_definition  = data_asset.add_batch_definition_whole_dataframe(name='customers_batch')
```

```python
#setup batch
batch = batch_definition.get_batch(    batch_parameters={
        "dataframe": df
    })
```

```python
#validate the data for the batch 
batch.validate(suite)
```

```python
print(batch_definition)
```

```python
#validation definition 
validation_definition = gx.ValidationDefinition(
    data=batch_definition,
    suite=suite,
    name="customer_quality_validation"
)
```

```python
#add the validation definition back to context
context.validation_definitions.add(validation_definition)
```

Insecpting failures

```python
bad_df = df.copy()
```

```python
bad_df.loc[1, "customer_id"] = bad_df.loc[0, "customer_id"]
bad_df.loc[2, "status"] = "pending"
bad_df.loc[3,'email'] = 'not-an-email'
```

```python
#get validation result
validation_result = validation_definition.run(
batch_parameters={"dataframe": bad_df}

)
```

```python
if validation_result.success:
    print("Validation PASSED")
else:
    print("Validation FAILED")
```

```python
for result in validation_result.results:
    if not result.success:
        print(result.expectation_config.type)
        print("Column:", result.expectation_config.kwargs.get("column"))
        print("Unexpected count:", result.result.get("unexpected_count"))
        print("Unexpected percent:", result.result.get("unexpected_percent"))
        print("Unexpected values:", result.result.get("partial_unexpected_list"))
        print()
```

```python
for result in validation_result.results:
    if not result.success:
        print(f'FAILED: {result.expectation_config.kwargs.get("column")} -> {result.result.get("unexpected_count")} unexpected values')
```
